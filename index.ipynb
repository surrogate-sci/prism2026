## Abstract

We're doing a series of experiments in order to look into whether we can use scientific assays and constructions in order to understand if models are causally using internal representations of scientific concepts in order to solve problems. We're doing this through using primarily the OpenADMET PXR dataset, and running both mech interp experiments and behavioural experiments, on scientific concepts like selectivity, potency, and dose-dependent responses.
